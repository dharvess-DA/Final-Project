## Stage 2 – Data Cleaning and Pre-processing

Handle missing values (impute or drop)

Handle duplicates

Treat outliers if required

Check skewness and apply transformations

Convert data types if needed

Feature transformations (date parts, derived fields if required for analysis)

```python
# Standerdizing the column name #
df.rename(columns={'STATE':'State','District Name':'District_Name',
                   'Market Name':'Market_Name','Price Date':'Price_Date'},inplace=True)
```

```python
# fill Categorical Columns #
df['State'] = df['State'].fillna(df['State'].mode()[0]) # return the more than one mode choose the highetst mode #
df['District_Name'] = df['District_Name'].fillna('Unknown')
df['Market_Name'] = df['Market_Name'].fillna('Unknown')
df['Commodity'] = df['Commodity'].fillna('Unknown')
df['Variety'] = df['Variety'].fillna('Unknown')
df['Grade'] = df['Grade'].fillna('Unknown')
```

```python
# fill numerical column Min_Price, Max_Price, Modal_Price columns has outliers .so we can use median #
df['Min_Price'] = df['Min_Price'].fillna(df['Min_Price'].median())
df['Max_Price'] = df['Max_Price'].fillna(df['Max_Price'].median())
df['Modal_Price'] = df['Modal_Price'].fillna(df['Modal_Price'].median())
```

```python
df['State'] = df['State'].str.strip()   # remove spaces
```

```python
df['State'] = df['State'].str.title()   # standard format
```

```python
# replace the inconsistance states name are mapping using dictionary #
state_mapping = {
    'Chhattisgarh': 'Chattisgarh',
    'Gao': 'Goa',
    'Jammu And Kashmir': 'Jammu & Kashmir',
    'Tamilnadu': 'Tamil Nadu'
}
df['State'] = df['State'].replace(state_mapping)
```

```python
df['Price_Date'] = pd.to_datetime(df['Price_Date'], errors='coerce')
df = df.dropna(subset=['Price_Date']) # Drop rows with no date
```

```python
df.count()
```

```python
# Handling Duplicates : Same State + Same District_Name + Same Market_Name + Same Commodity + Same Price_Date should be unique #
df.duplicated().sum()
```

```python
# using subset to identify the columns and keep first row and remove rest
df = df.drop_duplicates(subset=['State','District_Name','Market_Name','Commodity','Price_Date'],keep='first')
df

```

```python
# Check the Outliers using Box Plot #
price_cols = ['Min_Price', 'Max_Price', 'Modal_Price']
for col in price_cols:
    plt.figure()
    df.boxplot(column=col)
    plt.title(f'Boxplot of {col}')
    plt.show()
```

## **Interpretation of Outliers Box Plot(Before remove Outliers)**

. Min_price boxplot are concentrated in a narrow low range.However, a few extreme values increase the spread.These outliers may represent abnormal market conditions.

.The Max_Price boxplot distribution is highly skewed with a large number of outliers.This extreme outliers may reflect unusual market conditions.

.The Modal_Price boxplot  distribution is relatively stable and most value in a lower range,indicating consistent market pricing.



```python
# Removing outliers using IQR. In this Method Replace extreme values with boundary limits #
price_cols = ['Min_Price', 'Max_Price', 'Modal_Price']
for col in price_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    # outliers = df[(df[col] < lower) | (df[col] > upper)] #
    # print(f"{col} outliers:", len(outliers)) #
    df[col] = df[col].clip(lower, upper)
```

```python
# Check the Outliers after removing Outliers #
price_cols = ['Min_Price', 'Max_Price', 'Modal_Price']
for col in price_cols:
    plt.figure()
    df.boxplot(column=col)
    plt.title(f'Boxplot of {col}')
    plt.show()
```

# **Interpretation of Outliers Box Plot(After remove Outliers)**

. The Min_Price distribution box plot the median value approx.(2000) shows that the typical minimum price across the dataset

. The Max_price distribution box plot the median value approx (~2300) indicates that half of the observations have a Max_Price below this value and half above it. So, a typical maximum price lies around this level

.The Model_price distribution plot  the median value approx (~2100–2200) indicates that the most common (modal) price across markets is around this value.

.No strong outliers are visible, for the above all prices in which indicates that extreme prices from the overall distribution.



```python
# Convert all prices to numeric #
df['Min_Price'] = df['Min_Price'].astype(int)
df['Max_Price'] = df['Max_Price'].astype(int)
df['Modal_Price'] = df['Modal_Price'].astype(int)

```
